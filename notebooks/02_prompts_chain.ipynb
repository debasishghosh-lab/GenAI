# 02. Prompt Templates & LCEL Chains

In this notebook, we will learn:
1. How to create `PromptTemplate` and `ChatPromptTemplate`.
2. How to compose chains using LCEL (LangChain Expression Language): `prompt | llm | parser`.
3. How to invoke chains with variables.

```python
from dotenv import load_dotenv
from langchain_core.prompts import PromptTemplate, ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_google_genai import ChatGoogleGenerativeAI

# Load API keys from .env file
load_dotenv()
```

### 1. Basic `PromptTemplate`

```python
template = "Explain the concept of {topic} in simple terms for a {target_audience}."
prompt = PromptTemplate.from_template(template)

# Format the prompt
formatted_prompt = prompt.format(topic="Quantum Computing", target_audience="10 year old")
print(formatted_prompt)
```

### 2. `ChatPromptTemplate` for Chat Models

```python
chat_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are an expert technical tutor specializing in AI and Machine Learning."),
    ("user", "What is {concept} and how is it used in real-world applications?")
])

messages = chat_prompt.format_messages(concept="Retrieval Augmented Generation (RAG)")
print(messages)
```

### 3. Building an LCEL Chain (`prompt | llm | output_parser`)

```python
llm = ChatGoogleGenerativeAI(model="gemini-3.6-flash")
parser = StrOutputParser()

# Create the chain
chain = chat_prompt | llm | parser

# Invoke the chain
response = chain.invoke({"concept": "LangChain LCEL Pipes"})
print(response)
```
