```python
from dotenv import load_dotenv
load_dotenv()
```

```python
text="Kohli was the captain of the 2008 U19 World Cup winning team and was a crucial member of the Indian teams that won 2011 Cricket World Cup, 2013 Champions Trophy, 2024 T20 World Cup, and 2025 Champions Trophy. He plays for Royal Challengers Bengaluru in the Indian Premier League and for Delhi in domestic cricket. In 2013, Kohli was ranked number one in the ODI batting rankings. In 2015, he achieved the same in T20I cricket. In 2018, he was ranked number one in Test cricket, making him the only Indian to hold the top spot in all three formats. He is the first player to score 20,000 international runs in a decade. He was the Cricketer of the Decade for 2011 to 2020."
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(chunk_size=100, chunk_overlap=20)
chunks = text_splitter.split_text(text)
```

```python
print(chunks)
```

```python
for chunk in chunks:
    print(chunk)
```

Document_Splitter

```python
from langchain_community.document_loaders import PyPDFLoader
loader=PyPDFLoader("../data/Debasish_Ghosh_Resume.pdf")
docs=loader.load()
len(docs)
```

```python
splitter=RecursiveCharacterTextSplitter(chunk_size=200,chunk_overlap=50)
pdf_chunks=splitter.split_documents(docs)

len(pdf_chunks)
```

```python
for chunk in pdf_chunks:
    print(chunk.page_content)
    print()
```
