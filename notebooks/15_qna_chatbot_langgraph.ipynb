```python
from dotenv import load_dotenv
load_dotenv()
from pydantic import BaseModel
from langgraph.graph import START,END,StateGraph
from langchain_groq import ChatGroq
from langgraph.graph.message import add_messages
from langgraph.checkpoint.memory import InMemorySaver
from typing import Annotated
```

```python

class StateScheema(BaseModel):
    messages:Annotated[list,add_messages]
```

```python
graph=StateGraph(StateScheema)
```

```python
llm=ChatGroq(model="openai/gpt-oss-20b")

def LLM(state:StateScheema):
    res=llm.invoke(state.messages)
    state.messages=[res.content]
    return state


    
```

```python
Memory=InMemorySaver()
```

```python
graph.add_node("LLM",LLM)
graph.add_edge(START,"LLM")
graph.add_edge("LLM",END)
```

```python
final_graph=graph.compile(checkpointer=Memory)
```

```python
res=final_graph.invoke({"messages":[{"role":"user","content":"Hey am debasish "}]},
{"configurable":{"thread_id":"1"}}

)
print(res["messages"][-1].content)
```

```python
res=final_graph.invoke({"messages":[{"role":"user","content":"what is my name "}]},
{"configurable":{"thread_id":"1"}}

)
print(res["messages"][-1].content)
```
