# 03. Basic Memory with LLMs

In this notebook, we will learn:
1. How chat models process conversation history.
2. Using `HumanMessage`, `AIMessage`, and `SystemMessage`.
3. Using `MessagesPlaceholder` to pass ongoing conversation history into LCEL chains.

```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import HumanMessage, AIMessage, SystemMessage
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.output_parsers import StrOutputParser

# Load environment variables
load_dotenv()

# Initialize LLM
llm = ChatGoogleGenerativeAI(model="gemini-3.6-flash")
```

### 1. Manual Message List Management

```python
messages = [
    SystemMessage(content="You are a helpful AI assistant."),
    HumanMessage(content="Hi, my name is Debasish."),
    AIMessage(content="Hello Debasish! How can I assist you today?"),
    HumanMessage(content="What is my name?")
]

response = llm.invoke(messages)
print(response.content)
```

### 2. Conversation Memory using `MessagesPlaceholder` and LCEL

```python
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a polite assistant who remembers user preferences."),
    MessagesPlaceholder(variable_name="chat_history"),
    ("user", "{input}")
])

chain = prompt | llm | StrOutputParser()

# Maintain chat history state
history = [
    HumanMessage(content="My favorite programming language is Python."),
    AIMessage(content="That's awesome! Python is fantastic for Data Science and AI.")
]

# Ask a follow-up question
reply = chain.invoke({
    "chat_history": history,
    "input": "What language did I say I like?"
})

print(reply)
```
