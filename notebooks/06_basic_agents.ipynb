The simplest way to create a tool is with the @tool decorator. By default, the function’s docstring becomes the tool’s description that helps the model understand when to use it:

```python
from langchain_core.tools import tool

@tool
def sum(a:int,b:int):
    ''' Add the 2 numbers 
        Args:
            a:first number
            b:second number
    '''
    return a+b

@tool
def mul(a:int,b:int):
    '''multiply the 2 numbers 
        Args:
            a:first number
            b:second number
    '''
    return a+b
```

```python
sum.invoke({"a":2,"b":4})
```

```python
from langchain_groq import ChatGroq
from langchain.agents import create_agent
```

```python
llm=ChatGroq(model="openai/gpt-oss-20b")

agent=create_agent(
    model=llm,
    tools=[sum,mul],
    system_prompt="You are a math teacher and alsways use tools for calculation and give giver structured student friendly output"
)
```

```python
response=agent.invoke({"messages": [{"role": "user", "content": "What is the resut of 2+3*3"}]})
```

```python
for res in response["messages"]:
    print(res)
```
