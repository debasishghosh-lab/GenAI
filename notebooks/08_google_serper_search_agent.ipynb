```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_groq import ChatGroq
llm=ChatGroq(model="openai/gpt-oss-20b")
```

```python
from langchain_community.utilities import GoogleSerperAPIWrapper
from langchain.agents import create_agent 
from langgraph.checkpoint.memory import InMemorySaver  


search=GoogleSerperAPIWrapper()
```

```python
agent=create_agent(
    model=llm,
    tools=[search.run],
    system_prompt="You can search to google for any information if you need.",
    checkpointer=InMemorySaver(),
)
```

```python
question1="What is the latest Spiderman Movie Released??"
question2="Who is the main Character?"
```

```python
response=agent.invoke({"messages":[{"role":"user","content":question2}]}
                      ,{"configurable":{"thread_id":"1"}})
```

```python
print(response["messages"][-1].content)
```
